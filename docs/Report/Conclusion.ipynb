# Conclusion

This paper evaluated whether the Job-Ready Graduates Scheme produced enrolment shifts consistent with its intended pricing incentives, using a difference-in-differences estimator with UK and New Zealand enrolments as control groups across eleven disciplines from 2016 to 2024. The headline result is that the JRGS produced statistically significant effects in two of eleven disciplines: education enrolments increased by approximately 18.6% relative to the control countries following the 2021 reform, and management and commerce fell by approximately 21.6%. The remaining nine disciplines, including those combining the most acute labour market shortages with the most substantial fee reductions, produced null results.

Both significant results are consistent with a salience threshold interpretation of the Carter and Curry (2011) mechanism: the JRGS appears to have influenced enrolments only where fee changes moved disciplines outside the historical range of student contribution norms, below the prior floor in education and above the prior ceiling in management and commerce. Disciplines repositioned within the previously familiar bandwidth produced no detectable response. Both headline estimates are, however, materially constrained by robustness evidence. The education coefficient reverses sign under country-specific linear trends and is challenged by a significant UK wrong-group placebo attributable to independent domestic policy changes; the estimate is best treated as an upper bound rather than a point estimate of the JRGS effect. The management and commerce estimate collapses to near zero under NZ-only identification and is ranked last in permutation inference, indicating it cannot be credibly separated from broader post-2021 dynamics common to anglophone higher education systems.

Evaluated against the JRGS's stated objectives, the findings are broadly consistent with the Daly and Lewis (2020) characterisation of the scheme as unlikely to achieve meaningful enrolment reorientation, though the evidence here identifies a more specific failure mode. The policy appears to have produced visible enrolment shifts only at the extremes of its price structure while failing to generate the graduate supply response where the labour market need was greatest. Health, engineering, and information technology registered no significant enrolment change despite receiving the largest proportional fee reductions and exhibiting sustained shortage rates above 70% of relevant Skill Level 1 occupations. Entry into these vocational professions is governed by career commitment, prior subject investment, and intrinsic motivation that fee changes operating through a deferred repayment scheme are unlikely to overcome (Janke, 2020), suggesting the reductions functioned primarily as transfers to students who would have enrolled regardless rather than as supply-expanding incentives.

The policy implication is that price signals alone are insufficient to redirect graduate supply toward shortage-affected disciplines when enrolment decisions are governed primarily by non-financial factors and the financial cost itself is deferred to post-graduation income thresholds. Future evaluations would benefit from longer post-treatment panels as the 2021 cohort matures toward graduation, individual-level data capable of distinguishing compositional from headcount effects, and analysis of completion and employment outcomes to determine whether any demand shifts that did occur translated into the graduate supply the scheme intended to produce.

# Discussion

The results present a picture that neither fully validates nor fully dismisses the prior literature on JRGS effectiveness. Daly and Lewis (2020) characterised the scheme as a policy unlikely to shift enrolment patterns, a prediction consistent with the broader consensus that tuition price changes exert an inelastic effect on higher education demand at the market level (Langelett et al., 2015; Havranek et al., 2018). Nine of eleven disciplines produce null results, which is largely in line with that prediction. Two fields depart from it: education enrolments rose approximately 18.6% relative to the control countries and management and commerce fell approximately 21.6%, suggesting the JRGS may have produced real effects, but only under specific conditions that the inelastic demand model does not capture on its own.

The concentration of effects in the two disciplines with the largest absolute fee changes is consistent with the elasticity mechanism proposed by Carter and Curry (2011), which locates price responsiveness in higher education not at the aggregate level but between competing alternatives. The JRGS effectively reproduced the conditions of inter-institutional price competition within institutions, across disciplines. Before the reform, maximum student contributions were broadly compressed between &#36;6,500 and &#36;11,000 across all fields. After 2021 that range widened to &#36;4,000–&#36;15,000, with education falling conspicuously below the prior floor and management and commerce rising conspicuously above the prior ceiling. Engineering, health, IT, natural sciences, and environment all received fee cuts of between 20% and 45%, reducing their contributions to &#36;7,000–&#36;10,000 but keeping them within the previously familiar range, and none produced a statistically significant enrolment response. This pattern is consistent with the existence of a salience threshold: fee changes that keep a discipline within the bandwidth of what students have historically perceived as the normal cost of tertiary study may not register as a meaningful signal in the enrolment decision, while changes that move a field outside that range entirely are more likely to.

The absence of effects in health, engineering, and information technology is particularly striking given that these fields combine the most substantial fee reductions with the most acute documented labour market shortages; health shortage rates reached approximately 80% of Skill Level 1 occupations by 2023 and engineering peaked at 77% in the same year. If the JRGS was designed to increase graduate supply in fields where the labour market signal is strongest, these are precisely the disciplines where the policy should have shown the clearest results. Their null outcomes align with the university enrolment motivation literature: entry into vocational professions such as nursing, medicine, and engineering is shaped by career commitment, prior subject investment, and interpersonal factors that fee changes are unlikely to override (Janke, 2020). The JRGS fee reductions for these fields were also deferred in practice via the HECS-HELP scheme, reducing the immediate financial salience of the change at the point of enrolment decisions. Students not already drawn to those fields by non-financial motivations are unlikely to be redirected toward them by a reduction in deferred annual contribution, suggesting the fee cuts functioned primarily as a transfer to students who would have enrolled regardless rather than as an incentive that expanded the pool of graduates the labour market actually needed.

The robustness evidence substantially constrains the interpretation of both headline results. For education, the most important challenge is the UK wrong-group placebo: UK education enrolments were independently declining post-2021 due to domestic teacher training policy changes, meaning some portion of the estimated AUS–UK divergence reflects movement in the control group rather than a genuine Australian increase. When country-specific linear trends are incorporated into the specification, the education coefficient reverses sign from +0.171 to −0.043, a pattern corroborated by the pre-treatment trend data showing Australian education enrolments already growing faster than the controls from 2017 onward. The most conservative reading of the education result is therefore an upper bound on any genuine JRGS effect, with the true effect possibly smaller and not cleanly separable from a pre-existing trajectory. The management and commerce result faces a more fundamental structural problem: the NZ-only estimate collapses to +0.015, revealing that the pooled result is driven almost entirely by the AUS–UK comparison. With both wrong-group placebos producing statistically significant post-2021 movements and the permutation test placing the true Australian assignment last among all three possible treatment allocations (permutation p=1.00), the M&C estimate cannot be credibly isolated from whatever broader post-2021 dynamics were differentially affecting management enrolments across anglophone systems. It is better read as a descriptive finding consistent with the policy's direction than as an identified causal claim.

Evaluated against the JRGS's stated purpose of increasing graduate supply in high-priority, shortage-affected disciplines and reducing enrolment in fields with surplus supply, the results suggest the policy achieved its enrolment objectives in two disciplines but not the ones that matter most for the labour market problem it was designed to address. Education is a genuine shortage field and the observed increase, if real, is consistent with the policy's intent. Management and commerce is a field with persistently low shortage rates, and a decline in enrolments from a surplus-supply discipline is directionally appropriate. The null results across the remaining nine disciplines, however, include the fields with the most severe and sustained shortages: health, where shortage rates have remained above 70% of relevant occupations since 2022, and engineering and IT, which peaked above 75% before partially easing. A policy that succeeded in shifting students away from a low-shortage management field while failing to increase the supply of nurses, engineers, and IT graduates falls well short of the labour market correction it was presented as achieving. This is broadly consistent with the Daly and Lewis (2020) critique, though the evidence here points to a more specific failure mode: the JRGS appears to have been effective enough to produce visible enrolment shifts only at the extremes of its price structure, but insufficiently powerful to overcome the vocational, intrinsic, and structural factors that govern entry into the high-shortage professions where its effects were most needed.

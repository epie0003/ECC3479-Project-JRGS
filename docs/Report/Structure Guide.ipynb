## Section 3 - Empirical Strategy
**Target: ~2 pages**

This is the technical heart of the paper. Write it for a reader who knows econometrics but has not seen your data.

### 3.1 - Difference-in-Differences Setup

State the identifying variation:
- Treatment = Australia post-2021 (JRG enactment)
- Controls = UK and NZ, observed over the same period
- DiD recovers the ATE by subtracting the control group's post-period trend from Australia's change

### 3.2 - The Estimating Equation

Write the equation cleanly:

$$\log(E_{ct}) = \alpha + \beta_1 \cdot \text{Treated}_c + \beta_2 \cdot \text{NZ}_c + \beta_3 (\text{Treated}_c \times \text{Post}_t) + \sum_t \gamma_t \cdot \mathbf{1}_{[\text{year}=t]} + \varepsilon_{ct}$$

Define every term in a table (copy from `docs/Regression Analysis/DiD Modelling.ipynb`, Section 4). The coefficient of interest is **beta_3** - the DiD term.

Interpretation formula: *beta-hat_3 is the log-points difference between Australia's post-2021 enrolment change and the weighted average change in the UK and NZ. The approximate percentage effect is $(e^{\hat\beta_3} - 1) \times 100$.*

### 3.3 - Fixed Effects

- **Country FEs** (`treated`, `nz_dummy`): absorb all stable cross-country differences in enrolment levels - Australia being larger, NZ being smaller, cultural composition differences
- **Year FEs** (`C(year)`): absorb common shocks in each calendar year - global COVID disruption visible in all three countries simultaneously, global higher-education trends

### 3.4 - AUS-Specific COVID Controls

Explain that Australia's 2020 and 2021 pandemic disruptions (border closures, extended lockdowns) differ from UK/NZ. If AUS-specific COVID controls (`treated_covid2020`, `treated_covid2021`) are included in any of the per-discipline notebooks, mention them here. Note that year FEs partially handle this but cannot separate the AUS-specific component.

*(Check your per-discipline REG notebooks - if you included those controls, mention them explicitly here. If you dropped them in favour of COVID-sensitivity checks via the robustness section, note that instead.)*

### 3.5 - Standard Errors

This is a required justification. Write three sentences:

1. State that HC3 heteroskedasticity-robust standard errors are used throughout.
2. Justify: cross-country variance in absolute enrolment levels is substantial; homoskedasticity cannot be assumed.
3. Address clustering explicitly: *With G=3 country clusters, well below the ~42-cluster threshold required for cluster-robust SEs to be reliable (Bertrand, Duflo & Mullainathan 2004), clustering at the country level would produce degenerate standard errors. HC3 is the appropriate choice and is confirmed by Breusch-Pagan diagnostics in the Robustness section.*

### 3.6 - Identification: Parallel Trends Assumption

State the assumption formally and describe how it is tested:

> *The identifying assumption is that, absent the JRG, Australian enrolments would have followed a similar trend to the pooled UK and NZ after 2021. We assess this in two ways: (1) event-study plots of year-by-year DiD coefficients relative to a 2020 baseline - pre-treatment estimates (2016-2019) should not be statistically different from zero; (2) placebo tests with a fake structural break at 2019, estimated on the 2016-2020 pre-treatment window - an insignificant placebo coefficient supports the assumption.*

Reference the event study figures (already in REG notebooks) and placebo results (in Robustness Checks notebook).

## Section 6 - Discussion, Limitations & Conclusion
**Target: ~1.5 pages**

Three subsections: discussion of what the results mean -> limitations -> conclusion.

### 6.1 - Discussion

**1. What the significant results suggest about policy effectiveness:**
- Education beta=+0.171: approximately 18.6% enrolment increase. Consistent with the JRGS intention - a large fee reduction (to A$3,700 in 2021) was designed to attract more students into teacher training. Interpret in context: compare to base enrolment levels if possible.
- M&C beta=-0.243: approximately 21.6% enrolment decline. Consistent with the penalisation signal - fees rose sharply. Discuss whether this is the intended outcome (diverting graduates toward shortage occupations) or a concern about access to business education.

**2. Why most disciplines show no significant response:**
- Two possible explanations: (a) insufficient statistical power - N=18-27 is small; (b) genuine inelasticity - enrolment decisions may be driven by labour market expectations or personal interest more than fee levels
- Acknowledge both without resolving the ambiguity definitively

**3. Policy direction test outcome:**
- The two disciplines with the largest magnitude fee changes (Education down ~70%, M&C up ~100%) are the two that show significant movement - the others have smaller fee changes and no detectable response
- Interpret this as suggestive that the price signal *can* work at sufficient magnitude, but modest fee adjustments may be insufficient to shift enrolment behaviour

### 6.2 - Limitations

Each limitation in two sentences (what it is + what it implies for the results):

1. **Small panel:** Only N=27 or N=18 observations per discipline. This limits statistical power - most null results may reflect Type II error rather than a genuine absence of effect.

2. **Three-country panel / few controls:** UK and NZ are the only two control countries. Idiosyncratic shocks to either comparator could violate parallel trends even though we test for it with event studies.

3. **Parallel trends is untestable post-treatment:** Event study pre-trends are consistent with the assumption, but they do not prove it holds post-2021. Any factor affecting Australia differently from UK/NZ after 2021 - immigration patterns, labour market conditions, international student demand - would bias beta_3.

4. **Log-linear specification:** The model assumes a proportional effect of the policy on enrolments. If the true effect is heterogeneous, beta_3 is an average that may miss structural non-linearity.

5. **Discipline harmonisation:** Mapping AUS/UK/NZ subject classifications to 11 common categories involves judgment calls. Miscategorisation within a discipline would attenuate or distort the measured effect.

6. **Canada robustness check:** If the Canada 4-country panel check was run (from `data/clean/canada_enrollments_2016_2024.csv`), mention it here as a supplementary robustness check and note whether it confirms the main results.

### 6.3 - Conclusion

**Paragraph 1:** One-sentence restatement of the research question and approach.

**Paragraph 2:** The headline finding - Education up ~18.6% (p=0.035), M&C down ~21.6% (p=0.055), nine other disciplines statistically unchanged. Both significant results are directionally consistent with the policy's fee structure. This represents partial evidence that the JRGS price signals produced measurable enrolment shifts in the disciplines with the largest fee changes.

## Writing and Presentation Conventions

From Lecture 8 - apply these throughout.

### Regression Tables
- Standard errors always in parentheses below the coefficient
- Significance stars always defined in the notes (* p<0.05, dagger p<0.10)
- N, R-squared, and a description of fixed effects always included as table notes
- Never report raw p-values without also reporting the coefficient and SE
- Column headers describe the dependent variable and specification, not just a number

### Prose Conventions
- **Topic sentence -> column by column -> big-picture close** for any paragraph walking through a table
- State statistical significance with precision: *significant at the 5% level* or *marginally significant (p<0.10)* - not just *significant*
- Percentage effects from log coefficients: always compute $(e^{\hat\beta} - 1) \times 100$ and state it alongside the log-points
- Avoid: *the coefficient is large* - large relative to what? Always compare to a baseline or benchmark
- Avoid: *we can see from the table* - just state what the table shows

### Figures
- Every figure needs a numbered caption with a complete description (no reader should need to read the text to understand a standalone figure)
- Event study plots: label the treatment year visually (dashed vertical line); label the baseline year; show confidence intervals
- Forest/coefficient plots: label each specification on the y-axis; zero line must be visible; note that whiskers are 95% CI

# Literature Review

Price elasticity of demand for changes in tuition prices has historically been considered inelastic at the market level (Langelett, et al., 2015) and functionally 'close to zero' (Havranek, et al., 2018). Changes in overall pricing often lead to no significant change in enrolments, as regardless of tuition costs, most individuals seeking tertiary education will still pursue it. There are exceptions — such as between competing universities, where price differences are suggested to elicit a response from prospective students when deciding where to study (Carter & Curry, 2011) — however, this effect does not hold when market prices change uniformly, as with the JRGS. Instead, decisions to pursue tertiary education are suggested to be less dependent on pricing alone; factors such as availability, location, curriculum and discipline choice, and interpersonal relationships all contribute significantly to the decision-making process (Langelett, et al., 2015).

Applied to the JRGS, this literature suggests that market-level pricing changes are unlikely to affect overall enrolment, as the reform operated at a level where demand is inelastic. However, the specific targeting of high-priority and low-priority disciplines could potentially resemble the price elasticity observed between competing universities — where one institution is more expensive than another — but applied instead to disciplines competing within the same institution. This introduces its own complications. Literature surrounding university enrolment motivations is in tension with the use of price signals to direct disciplinary choice. Enrolment decisions are shaped by balancing intrinsic and extrinsic motivations; by radically shifting prices in either direction, the JRGS places a premium on extrinsic factors — financial costs — which may crowd out intrinsic factors such as personal interest or enjoyment. Intrinsic motivation is important for sustaining long-term persistence in study, and where it is displaced, students are more likely to struggle with motivation, satisfaction, and completion of their degree (Janke, 2020). Utilising price signals to tilt the balance toward extrinsic factors penalises students for prioritising intrinsic motivations deemed 'low-priority'. If the JRGS does affect enrolment, it may consequently harm student welfare for those not intrinsically motivated toward the designated high-priority disciplines.

Specific literature evaluating the JRGS is limited, though the available findings are consistent with an inelastic demand model and are broadly critical of the policy. Prior pricing changes — including adjustments to HECS/HELP contribution rates — did not produce meaningful changes in enrolments despite their magnitude being smaller than those introduced by the JRGS (Daly & Lewis, 2020). This can be partially attributed to the structure of the HECS/HELP scheme itself, which permits students to defer fees until their income exceeds the repayment threshold, potentially reducing the perceived financial cost of a degree at the point of selection. It can also be attributed to the variation in personal and preference factors that are suggested to hold greater influence over tertiary enrolment decisions than price alone (Daly & Lewis, 2020).

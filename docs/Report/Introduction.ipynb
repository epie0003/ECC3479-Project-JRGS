# Introduction

In the lead-up to a set of education reforms introduced in 2020, there was a perceived mismatch between popular fields of study and areas of job shortage in the Australian economy. The federal government's response was the Job-Ready Graduates Scheme (JRGS), implemented in 2021, which restructured domestic undergraduate tuition fees to incentivise high-priority disciplines (education, health, agriculture, engineering, and IT) through increased commonwealth funding and reduced student contributions, while discouraging low-priority disciplines such as management & commerce, law, arts, and social sciences through the inverse. Before the JRGS, maximum student contributions across disciplines ranged from roughly AUD 6,500 to AUD 11,000 per annum; following the reform, priority disciplines fell to approximately AUD 4,000–9,500 while non-priority disciplines rose to AUD 11,500–16,500, producing a spread of up to AUD 12,500 within a single year of study. Literature on tuition price elasticity suggests demand is inelastic at the market level but elastic when comparing variations across competing institutions (Carter & Curry, 2011), with other factors including location, curriculum, and interpersonal relationships also shaping enrolment decisions (Langelett et al., 2015). Existing evaluations of the JRGS have likened it to 'a misguided arrow missing its target', arguing it is 'unlikely to have a significant effect on student choices' and that disciplinary selection is more heavily driven by socio-economic background, career guidance, and occupational expectations not addressed by the scheme (Daly & Lewis, 2020). However, these evaluations do not examine whether the policy may have succeeded selectively across specific disciplines, a gap this paper addresses by providing the first discipline-level evaluation of the JRGS using a cross-national difference-in-differences design.

This paper tests whether the JRGS produced enrolment shifts consistent with its intended pricing incentives, and whether any observed shifts align with the policy design. The approach uses a difference-in-differences estimator with UK and New Zealand (NZ) university enrolments as treatment-absent control groups. Both represent appropriate comparators: they share similar higher education structures, income-contingent loan schemes, and broadly comparable enrolment patterns, yet neither underwent a discipline-specific fee reform during the study period. The UK's domestic fee cap (£9,250 per annum) has been frozen uniformly across all subjects since 2017, meaning UK enrolment variation by discipline reflects demand and supply factors rather than price incentives. NZ's Fees Free policy (effective 2018) subsidised the first year of study universally across all disciplines, placing it fully in the pre-treatment window and rendering it orthogonal to discipline-level identification. University enrolment panel data spanning 2016–2024 across eleven disciplines for all three countries is used, alongside panel data on Australian university funding clusters and job shortage indicators. The estimating equation applies two-way fixed effects (country and year) with HC3 standard errors, justified by the small number of country clusters (G=3).

The findings suggest the JRGS produced a detectable enrolment response in two disciplines: education enrolments increased by approximately 18.6% relative to control countries after 2021 (p=0.035, significant at 5%), while management & commerce fell by approximately 21.6% (p=0.055, significant at 10% only). The remaining nine disciplines showed no statistically significant change. While the two disciplines with the largest fee changes moved in the direction the JRGS intended, consistent with students responding to price signals, these results alone cannot establish whether this reflects genuine demand elasticity or concurrent confounding factors. The null effects across the remaining disciplines suggest the scheme had limited broader impact in realigning enrolment with labour market priorities.

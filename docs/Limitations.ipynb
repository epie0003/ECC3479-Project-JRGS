# Limitations

This document catalogues the principal limitations of the analysis, organised by category.
Each limitation is stated concisely for use in the report's limitations section.


## 1. Data Limitations

### 1.1 Uneven panel windows across disciplines

UK HESA subject classifications changed in 2019/20, making pre-2019 UK data
irreconcilable with later figures for five disciplines (Natural & Physical Sciences,
Environment & Related, Society & Culture, Others, Architecture & Building). These
disciplines enter the panel in 2019, leaving only one pre-policy observation before
the JRG commenced in 2021. With a single pre-treatment year, pre-existing differential
trends between countries cannot be distinguished from a policy effect; estimates for
these five disciplines carry materially higher uncertainty than those with the full
2016–2024 panel.

### 1.2 Calendar misalignment between countries

Australian enrolment data is recorded by calendar year; UK HESA data is reported by
academic year (e.g., 2020/21). All UK observations are assigned to their start year
for comparability, introducing an approximate six-month misalignment. This is a
standard harmonisation decision for cross-country education panels but means that
year-on-year comparisons capture slightly different academic cohorts in each country.

### 1.3 Aggregate discipline-level data obscures sub-field composition

The analysis uses total enrolments at the broad field level (e.g., all of "Health" or
all of "Management & Commerce"). If the JRG caused students to shift between sub-fields
within the same broad category — for example, from law to social work within Society &
Culture, or between nursing and allied health — those substitution effects would not
appear in the aggregate counts. The true policy response may be more varied across
sub-fields than these estimates suggest.

### 1.4 Short post-policy observation window

The panel covers only 2021–2024 — four years of post-JRG data. Enrolment responses to
fee changes are expected to build over multiple cohorts, since students commit to
multi-year programmes 1–2 years in advance. The estimates in this analysis likely
understate the eventual long-run effect, particularly for discouraged fields where
price-sensitivity may accumulate across cohorts.


## 2. Identification Limitations

### 2.1 Small number of comparison countries (G = 3)

Australia is compared against only the United Kingdom and New Zealand. With three
country groups, the parallel trends assumption cannot be verified statistically — there
is no third comparator against which to check that the control countries are a credible
counterfactual for Australia. If either the UK or NZ experienced its own discipline-specific
enrolment shocks after 2021 (e.g., UK domestic tuition policy changes, NZ cost-of-living
pressures), those shocks are absorbed into the estimated JRG effect and cannot be separated
out. The robustness checks in Section 2.3 and 2.6 of the robustness notebook document
several cases where control-country-specific movements contaminate the estimate.

### 2.2 COVID-19 affected countries asymmetrically

Australia's international border closures ran from early 2020 to early 2022 — longer than
the UK or NZ — creating an asymmetry in the 2020 baseline that the event study rests on.
In disciplines with high international student shares (IT, Engineering, Management & Commerce),
the post-2021 recovery in AUS enrolments may partly reflect reopening rather than a pure
JRG policy effect. Year fixed effects absorb the global COVID shock visible in all three
countries; the AUS-specific border-closure component is partially controlled via additional
COVID interaction terms in the robustness checks but cannot be fully separated from the
fee-reform signal.

### 2.3 Single treated unit

Australia is the only treated country. With a single treatment unit, the DiD estimate
cannot be separated from any other AUS-specific shock that happened to coincide with the
2021 JRG reform. Any concurrent AUS policy change affecting higher education enrolments
(domestic demand shifts, university marketing, funding system changes unrelated to JRG fee
rates) would be attributed to the JRG by the DiD design.

### 2.4 NZ disproportionate influence given small scale

NZ enrolments are typically 5–10× smaller than Australia's across most disciplines. In
smaller categories, a shift of a few hundred NZ students driven by local factors can
meaningfully shift the estimated control-group trend and inflate or deflate the apparent
AUS policy effect. The pooled-control model does not weight countries by enrolment scale,
giving NZ disproportionate influence in the within-year comparison relative to its size.


## 3. Robustness and Inference Limitations

### 3.1 Degenerate clustering with G = 3

Standard cluster-robust standard errors require a minimum of approximately 42 clusters
for the asymptotic approximation to be reliable (Angrist & Pischke 2008; Cameron &
Miller 2015). With only G = 3 country clusters, cluster-robust SEs cannot be used.
HC3 heteroskedasticity-robust standard errors are used throughout as the appropriate
small-sample alternative, justified via a Moulton factor analysis showing within-cluster
correlation is not the dominant source of variance. This is a structural constraint of
any three-country panel design and cannot be resolved without additional comparison
countries.

### 3.2 Permutation inference cannot achieve conventional significance thresholds

With G = 3, the exact permutation test can only enumerate three possible treatment
assignments. The minimum achievable permutation p-value is 1/3 ≈ 0.33 — meaning even
a genuine effect cannot achieve p < 0.05 under permutation. The permutation results
confirm that for Management & Commerce, both placebo assignments produce larger absolute
coefficients than the real AUS assignment (permutation p = 1.00), and for Education the
real assignment ranks second (permutation p = 0.67). These results reinforce the caution
flags from the wrong-group placebo tests rather than adding new significance information.

### 3.3 Education: parallel trends assumption is not robust to country-specific trends

When country-specific linear trend terms are added to the main specification
(year_c:treated, year_c:nz_dummy), the Education coefficient reverses sign — from
+0.171 (main) to −0.043 (with trends). This indicates that AUS Education enrolments
were on a diverging linear trajectory relative to the UK throughout the sample period,
and part of the post-2021 DiD estimate may reflect a continuation of this pre-existing
differential rather than a genuine JRG discontinuity. The Education result should
therefore be treated as an upper bound on the pure fee-reform effect.

### 3.4 Management & Commerce result is not robustly identified

The M&C coefficient (β = −0.243, p = 0.055) fails four compounding robustness checks:
both wrong-group placebos are statistically significant (UK: p < 0.001, NZ: p = 0.045),
indicating global post-COVID M&C reallocation across anglophone systems; the result
collapses to insignificance when two UK pre-treatment observations are dropped (Cook's
D influence check); and the permutation test places the real AUS assignment last among
all three possible assignments. The directional finding is consistent with the JRG
discouraged-field fee increase, but the design cannot cleanly separate this from
concurrent global demand shifts in international business and commerce enrolments.

### 3.5 Wrong-outcome placebo not implementable

A third flavour of placebo test — using an outcome the JRG should not affect, such as
domestic student enrolments or academic staff headcounts — would strengthen the
identification argument by ruling out general AUS higher-education trends. This test
requires additional data not currently in the project and cannot be conducted here.


## 4. Summary

| Category | Key limitation | Severity |
|----------|---------------|---------|
| Data | Uneven panel windows (5 disciplines N=18, one pre-JRG year) | High |
| Data | Short post-policy window (2021–2024 only) | Medium |
| Data | Aggregate discipline data hides sub-field composition | Medium |
| Identification | G=3 comparison countries — parallel trends unverifiable | High |
| Identification | Single treated unit — concurrent AUS shocks uncontrolled | High |
| Identification | COVID asymmetry between countries | Medium |
| Inference | Cluster-robust SEs infeasible; HC3 used throughout | Medium |
| Inference | Education sign reverses with country-specific trends | High |
| Inference | M&C fails four robustness checks; result not cleanly identified | High |
| Inference | Permutation minimum p = 1/3; conventional thresholds unreachable | Low |

Limitations rated **High** directly affect the causal interpretation of the two headline
results and should be stated explicitly in the conclusion. Those rated **Medium** affect
precision or external validity. **Low**-rated limitations are structural constraints of
the design that are documented but do not change the substantive conclusions.
